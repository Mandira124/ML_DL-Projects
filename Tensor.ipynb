#**Theoretical Background**

A tensor is a way to store data in multiple dimensions. It can be a single number (scalar), a list (vector), or a table (matrix). In machine learning, all data is stored as tensors.

Reshaping is used to change the structure of data without changing its values. This is important when preparing input for models.

Broadcasting allows operations between tensors of different shapes by automatically expanding them. This helps avoid loops and makes computation faster.

Matrix multiplication is a key operation in machine learning. It is used in neural networks to combine inputs with weights.

Formula for matrix multiplication:

C = A × B

Each element is calculated by multiplying rows and columns.

Eigenvalues represent important directions in data. They are used in techniques like PCA (Principal Component Analysis).

Entropy is a measure of uncertainty in probability. Higher entropy means more randomness.

#**IMPLEMENTATION**

Tensor Creation, Reshaping, Broadcasting

```python
import numpy as np
import torch

print("Tensor Creation")

# NumPy tensor
np_tensor = np.array([[1, 2, 3], [4, 5, 6]])
print("NumPy Tensor:\n", np_tensor)

# PyTorch tensor
torch_tensor = torch.tensor([[1, 2, 3], [4, 5, 6]])
print("PyTorch Tensor:\n", torch_tensor)


print("Reshaping")

x = np.arange(6)
reshaped = x.reshape(2, 3)
print("Original:", x)
print("Reshaped:\n", reshaped)


print("Broadcasting")

a = np.array([[1, 2, 3]])
b = np.array([[1], [2], [3]])

broadcast_result = a + b
print("Broadcasting Result:\n", broadcast_result)
```

Matrix Multiplication & Eigenvalues

```python
print("Matrix Multiplication\n")

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

np_result = np.dot(A, B)
print("NumPy Multiplication:\n", np_result)


print("Eigenvalues")

eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

Matrix Multiplication (Manual vs NumPy vs PyTorch)

```python
print("Manual Matrix Multiplication")

# Manual multiplication
manual_result = [
    [A[0][0]*B[0][0] + A[0][1]*B[1][0],
     A[0][0]*B[0][1] + A[0][1]*B[1][1]],

    [A[1][0]*B[0][0] + A[1][1]*B[1][0],
     A[1][0]*B[0][1] + A[1][1]*B[1][1]]
]

print("Manual Result:\n", manual_result)


print("PyTorch Multiplication")

A_t = torch.tensor(A, dtype=torch.float32)
B_t = torch.tensor(B, dtype=torch.float32)

torch_result = torch.matmul(A_t, B_t)
print("PyTorch Result:\n", torch_result)
```

Numerical Instability / Precision Experiment

```python
print(" Numerical Instability ")

a = 1e10
b = 1e-10

result = (a + b) - a

print("Expected:", b)
print("Actual:", result)
```
